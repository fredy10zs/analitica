## Desarrollo

```python

from pathlib import Path
import sys

# Directorio 
CARPETA_ACTIVIDAD = Path.cwd()

CARPETA_DATOS = CARPETA_ACTIVIDAD.parent / "datos_csv_json"
CARPETA_REPORTES = CARPETA_ACTIVIDAD / "reportes"

# Archivos originales
RUTA_CSV_ORIGINAL = CARPETA_DATOS / "FastFoodRestaurants.csv"
RUTA_JSON_ORIGINAL = CARPETA_DATOS / "countries.json"

# Nuevos archivos
RUTA_CSV_MODIFICADO = CARPETA_DATOS / "FastFoodRestaurants_modificado.csv"
RUTA_JSON_MODIFICADO = CARPETA_DATOS / "countries_modificado.json"
RUTA_RESUMEN = CARPETA_REPORTES / "resumen_csv_json.txt"

URL_CSV = (
    "https://huggingface.co/datasets/"
    "JasonZhouTI/Fast_Food_Restaurants/resolve/main/"
    "FastFoodRestaurants.csv?download=true"
)

URL_JSON = (
    "https://huggingface.co/datasets/"
    "false-facts-finetuning/country-capitals/blob/main/countries.json"
)


#importar las clases 
CARPETA_CLASES = CARPETA_ACTIVIDAD.parent / "clases_po"
if str(CARPETA_CLASES.parent) not in sys.path:
    sys.path.insert(0, str(CARPETA_CLASES.parent))

CARPETA_REPORTES.mkdir(parents=True, exist_ok=True)

print("Carpeta de trabajo:", CARPETA_ACTIVIDAD)
print("CSV original existe:", RUTA_CSV_ORIGINAL.exists())
print("JSON original existe:", RUTA_JSON_ORIGINAL.exists())
print("Carpeta de reportes:", CARPETA_REPORTES)
```

```python

from pathlib import Path

CARPETA_PROYECTO = Path.cwd().parents[1]

for nombre in ["FastFoodRestaurants.csv", "countries.json"]:
    encontrados = list(CARPETA_PROYECTO.rglob(nombre))
    print(f"\n{nombre}:")
    if encontrados:
        for ruta in encontrados[:10]:
            print(" ", ruta)
    else:
        print("  No se encontró dentro del proyecto.")
```

```python

from clases_po.manejador_csv import ManejadorCSV
from clases_po.manejador_json import ManejadorJSON

print("La clase ManejadorCSV fue importada correctamente.")
print("La clase ManejadorJSON fue importada correctamente.")
```

```python

import csv
import json

with open(RUTA_CSV_ORIGINAL, "r", encoding="utf-8", newline="") as archivo:
    lector = csv.reader(archivo)
    encabezados_csv = next(lector, [])
    primeras_filas_csv = [next(lector, None) for _ in range(3)]

print("CSV Original")
print("Columnas:", encabezados_csv)
print("Primeras filas:")
for fila in primeras_filas_csv:
    if fila is not None:
        print(fila)

with open(RUTA_JSON_ORIGINAL, "r", encoding="utf-8") as archivo:
    datos_json_original = json.load(archivo)

print("\nJson Original")
print("Tipo de dato:", type(datos_json_original).__name__)

if isinstance(datos_json_original, dict):
    print("Claves principales:", list(datos_json_original.keys())[:10])
elif isinstance(datos_json_original, list):
    print("Cantidad de elementos:", len(datos_json_original))
    print("Primer elemento:")
    print(json.dumps(datos_json_original[0] if datos_json_original else None,
                     ensure_ascii=False, indent=2)[:1500])
else:
    print("Contenido:", str(datos_json_original)[:1500])
```

```python

# Crear una instancia 
manejador_json = ManejadorJSON(RUTA_JSON_ORIGINAL)

datos_json = manejador_json.importar()

print("Antes de la modificación")
print("País:", datos_json[0]["name"])
print("Nombre visible:", datos_json[0]["display_name"])
print("Total de países:", len(datos_json))

# Modificar los datos en memoria
manejador_json.modificar()

print("\nDespués de la modificación")
print("País:", datos_json[0]["name"])
print("Nombre visible:", datos_json[0]["display_name"])
print("Cambio realizado:", manejador_json.modificacion)
print("Total de países:", len(datos_json))
```

```python
import csv
import json

# Verificar 
with open(RUTA_CSV_MODIFICADO, "r", encoding="utf-8", newline="") as archivo:
    lector_csv = csv.DictReader(archivo)
    primera_fila_csv = next(lector_csv, None)

print("Verificación del csv ")
print("Nombre guardado:", primera_fila_csv["name"] if primera_fila_csv else "Sin registros")

# Verificar el JSON guardado
with open(RUTA_JSON_MODIFICADO, "r", encoding="utf-8") as archivo:
    datos_json_guardados = json.load(archivo)

print("\nVerificación del json")
print("País:", datos_json_guardados[0]["name"])
print("Nombre visible guardado:", datos_json_guardados[0]["display_name"])
print("Total de países:", len(datos_json_guardados))

# Confirmar que los originales no fueron modificados
with open(RUTA_JSON_ORIGINAL, "r", encoding="utf-8") as archivo:
    datos_json_originales_verificacion = json.load(archivo)

print("\nComparación")
print("Nombre visible original:", datos_json_originales_verificacion[0]["display_name"])
```

```python

from clases_po.generador_resumen_txt import GeneradorResumenTXT

print("Clase GeneradorResumen fue importada correctamente.")
```

```python
#Resumen de los datos
resultado_csv = {
    "archivo_origen": str(RUTA_CSV_ORIGINAL),
    "modificacion": manejador_csv.modificacion,
    "archivo_salida": str(RUTA_CSV_MODIFICADO),
    "registros": len(datos_csv),
    "clase": "ManejadorCSV",
    "metodos": "importar(), modificar(), guardar()",
}

resultado_json = {
    "archivo_origen": str(RUTA_JSON_ORIGINAL),
    "modificacion": manejador_json.modificacion,
    "archivo_salida": str(RUTA_JSON_MODIFICADO),
    "registros": len(datos_json),
    "clase": "ManejadorJSON",
    "metodos": "importar(), modificar(), guardar()",
}

print("Resultados csv")
for clave, valor in resultado_csv.items():
    print(f"{clave}: {valor}")

print("\nResulytados json")
for clave, valor in resultado_json.items():
    print(f"{clave}: {valor}")
```

```python
# Crear el generador del resumen
generador = GeneradorResumenTXT(
    url_csv=URL_CSV,
    url_json=URL_JSON,
    resultado_csv=resultado_csv,
    resultado_json=resultado_json,
)

ruta_resumen_generado = generador.generar(str(RUTA_RESUMEN))

print("El resumen fue generado correctamente.")
print("Ruta:", ruta_resumen_generado)
print("El archivo existe:", RUTA_RESUMEN.exists())
```
