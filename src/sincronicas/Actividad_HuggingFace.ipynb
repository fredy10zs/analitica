# Actividad: Carga y transformación de datos con Hugging Face

## Objetivo

Cargar un conjunto de datos desde Hugging Face utilizando Python y Pandas, realizar una transformación de los datos y generar un reporte descriptivo del DataFrame resultante.

## Dataset

- **Nombre:** Fast Food Restaurants
- **Repositorio:** `JasonZhouTI/Fast_Food_Restaurants`
- **Origen:** Hugging Face

## Transformación propuesta

Conservar únicamente los restaurantes que tengan una página web registrada, para trabajar con información disponible y generar un reporte de los datos transformados.

```python

#preparar importaciones

import sys
from pathlib import Path

# Encontrar la raíz del proyecto
ROOT = Path.cwd().resolve()

while ROOT != ROOT.parent and not (ROOT / "src" / "clases_po").exists():
    ROOT = ROOT.parent

# Permitir importar las clases del proyecto
sys.path.insert(0, str(ROOT / "src"))

from clases_po import HFDatasetLoader, Reportes

import pandas as pd

print("Raíz del proyecto:", ROOT)
print("Importaciones realizadas correctamente")
```

```python
#Cargar el dataset desde Hugging Face


loader = HFDatasetLoader(
    dataset_name="JasonZhouTI/Fast_Food_Restaurants",
    split="train",
)

df_original = loader.load()

print("Dataset cargado correctamente")
print("Dimensiones:", df_original.shape)
print("Columnas:", df_original.columns.tolist())

df_original.head()
```

```python
#Revisar los valores nulos 

print("Dimensiones del dataset original:", df_original.shape)

print("\nValores nulos por columna:")
print(df_original.isna().sum())

print("\nTotal de filas duplicadas:", df_original.duplicated().sum())
```

```python
#Transformar el dataframe 


df_transformado = df_original[
    df_original["websites"].notna()
    & df_original["websites"].astype("string").str.strip().ne("")
].copy()

print("Filas originales:", len(df_original))
print("Filas después de la transformación:", len(df_transformado))
print("Restaurantes excluidos:", len(df_original) - len(df_transformado))

df_transformado.head()
```

```python
#Importar la clase Reportes


import sys
from pathlib import Path

ROOT = Path.cwd().resolve()

while ROOT != ROOT.parent and not (ROOT / "src" / "clases_po").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT / "src"))

from clases_po import Reportes

print("Clase Reportes importada correctamente")
```

```python
#Generar el reporte del DataFrame transformado


reportes = Reportes()

texto_reporte = reportes.generar_reporte(df_transformado)

print(texto_reporte)
```

```python
#Guardar el reporte en un archivo independiente


ruta_reporte = ROOT / "src" / "sincronicas" / "reportes" / "fast_food_restaurants_transformado.txt"

reportes.guardar_reporte(df_transformado, ruta_reporte)

print("Reporte guardado en:", ruta_reporte)
```

```python
#Verificar que el archivo existe


print("¿Existe el archivo?", ruta_reporte.exists())

print("\nPrimeras líneas del reporte:\n")
print("\n".join(ruta_reporte.read_text(encoding="utf-8").splitlines()[:12]))
```

```python
#Comprobar los resultados finales


print("RESUMEN FINAL DE LA ACTIVIDAD")
print("=" * 40)
print("Filas originales:", len(df_original))
print("Filas transformadas:", len(df_transformado))
print("Columnas:", len(df_transformado.columns))
print("Valores nulos en websites:", df_transformado["websites"].isna().sum())
print("Filas duplicadas:", df_transformado.duplicated().sum())
print("Reporte guardado:", ruta_reporte.exists())
```
